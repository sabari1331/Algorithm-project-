```python
from scipy import stats
sample = [2500,1800,5000,69000,109000,5000,3300,3500,5000]
sample_mean = np.mean(sample)
ci = stats.t.interval(
    confidence = 0.34,
    df=len(sample)-1,
    loc = sample_mean,
    scale=stats.sem(sample)
)

print(sample_mean)
print("34% Confidence Interval", ci)
```

```python
from scipy.stats import norm
sample=[2500,1800,5000,69000,109000,5000,3300,3500,5000]
sample_mean = np.mean(sample)
sample_mean
```

```python
sample_std = np.std(sample,ddof=1)
sample_std
```

```python
n=len(sample)
n
```

```python
z=1.56
```

```python
margin=z*(sample_std/np.sqrt(n))
margin
```

```python
upper_ci =(sample_mean+margin)
lower_ci =(sample_mean-margin) 

print(upper_ci)
print(lower_ci)
```

```python
print("Lower CI",round(lower_ci,2))
```
