```python
import pandas as pd
```

```python
d=pd.read_csv(r"C:\Users\sivas\Downloads\Bank-Customers.csv")
d
```

```python
x=d[["Earning","Credit Score"]]
x
```

```python
import scipy.cluster.hierarchy as sch
```

```python
import matplotlib.pyplot as plt
```

```python
s=sch.dendrogram(sch.linkage(x))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
cm=AgglomerativeClustering(n_clusters=4,linkage="ward")
y_pred = cm.fit_predict(x)
```

```python
plt.scatter(x.iloc[:,0],x.iloc[:,1],c=y_pred,cmap="rainbow")
plt.show()
```
