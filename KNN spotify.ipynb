```python
x=d[['danceability','tempo','loudness','Valence','energy']]
x
```

```python
from sklearn.preprocessing import StandardScaler 
```

```python
ss=StandardScaler()
```

```python
x=ss.fit_transform(x)
x
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
model=NearestNeighbors()
model.fit(x)
```

```python
x=['danceability','tempo','loudness','Valence','energy']

def rec (song):
    s_index=d[d["song_name"]==song].index[0]

    query_point=d.loc[[s_index],x].values

    distance,indices=model.kneighbors(query_point)
    return d.loc[indices[0]]["song_name"]


print(rec("Angalamma"))
```
